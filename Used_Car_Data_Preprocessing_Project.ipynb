```python
import pandas as pd
import numpy as np

data = {
    'Make': [
        'Maruti', 'Hyundai', 'Honda', 'Toyota', 'BMW',
        'Audi', 'Maruti', 'Hyundai', None, 'Honda',
        'Toyota', 'BMW', 'Audi', 'Maruti', 'Hyundai'
    ],

    'Model': [
        'Swift', 'i20', 'City', 'Fortuner', 'X5',
        'A4', 'Baleno', 'Creta', 'Verna', None,
        'Innova', 'X3', 'A6', 'Alto', 'Venue'
    ],

    'Price': [
        500000, 700000, 900000, 3500000, 6000000,
        4500000, -5000, 800000, 650000, 1200000,
        np.nan, 5500000, 99999999, 400000, 750000
    ],

    'Mileage': [
        '50,000 km', '30,000 km', '45,000 km',
        '20,000 km', '15,000 km', '25,000 km',
        None, '40,000 km', '35,000 km',
        '60,000 km', '55,000 km', '18,000 km',
        '22,000 km', '70,000 km', '28,000 km'
    ],

    'Year': [
        2018, 2020, 2019, 2021, 2022,
        2023, 2025, 2030, 2017, 2016,
        2018, 2020, 2021, 1995, np.nan
    ],

    'Fuel_Type': [
        'Petrol', 'Diesel', 'Petrol', 'Diesel', 'Petrol',
        'Diesel', 'Petrol', None, 'Petrol', 'Diesel',
        'Petrol', 'Diesel', 'Petrol', 'Petrol', 'Diesel'
    ],

    'Transmission': [
        'Manual', 'Manual', 'Automatic', 'Automatic', 'Automatic',
        'Automatic', 'Manual', 'Manual', 'Automatic', 'Manual',
        None, 'Automatic', 'Automatic', 'Manual', 'Manual'
    ],

    'Condition': [
        'Good', 'Excellent', 'Good', None, 'Excellent',
        'Good', 'Fair', 'Good', 'Excellent', None,
        'Good', 'Excellent', 'Good', 'Fair', 'Excellent'
    ],

    'Engine_Size': [
        1.2, 1.4, 1.5, 2.8, 3.0,
        np.nan, 1.2, 1.5, 1.4, 1.5,
        2.4, 3.0, 2.0, 0.8, 1.0
    ]
}

df = pd.DataFrame(data)

print("Raw Dataset Created Successfully!")
df
```

```python
df.to_csv('used_cars_raw.csv', index=False)

print("Raw dataset saved successfully!")
```

```python
print("First 5 Rows:")
display(df.head())

print("\nDataset Information:")
df.info()

print("\nStatistical Summary:")
display(df.describe(include='all'))
```

```python
print("Missing Values in Each Column:")

missing_values = df.isnull().sum()

print(missing_values)
```

```python
df['Mileage'] = df['Mileage'].astype(str)

df['Mileage'] = df['Mileage'].str.replace(' km', '', regex=False)

df['Mileage'] = df['Mileage'].str.replace(',', '', regex=False)

df['Mileage'] = pd.to_numeric(
    df['Mileage'],
    errors='coerce'
)

print("Mileage column cleaned!")
display(df[['Mileage']].head())
```

```python
current_year = 2024

df.loc[
    (df['Year'] > current_year) |
    (df['Year'] < 1990),
    'Year'
] = np.nan

print("Invalid Year values converted to missing values!")
display(df[['Year']])
```

```python
df.loc[df['Price'] <= 0, 'Price'] = np.nan

print("Invalid prices fixed!")
display(df[['Price']])
```

```python
missing_percentages = df.isnull().sum() / len(df) * 100

print("Missing Data Percentage:")
print(missing_percentages)
```

```python
df['Mileage'] = df['Mileage'].fillna(
    df['Mileage'].median()
)

df['Price'] = df['Price'].fillna(
    df['Price'].median()
)

df['Engine_Size'] = df['Engine_Size'].fillna(
    df['Engine_Size'].median()
)

df['Year'] = df['Year'].fillna(
    df['Year'].median()
)

print("Numerical missing values handled!")
df['Condition'] = df['Condition'].fillna(
    df['Condition'].mode()[0]
)

df['Fuel_Type'] = df['Fuel_Type'].fillna(
    df['Fuel_Type'].mode()[0]
)

df['Transmission'] = df['Transmission'].fillna(
    df['Transmission'].mode()[0]
)

print("Categorical missing values handled!")
df = df.dropna(subset=['Make', 'Model'])

print("Rows with missing Make or Model removed!")
print("New Dataset Shape:", df.shape)
print("Remaining Missing Values:")

print(df.isnull().sum())

```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8, 5))

sns.boxplot(x=df['Price'])

plt.title("Price Before Outlier Treatment")

plt.show()
```

```python
def cap_outliers_iqr(dataframe, column):

    Q1 = dataframe[column].quantile(0.25)

    Q3 = dataframe[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR

    upper_bound = Q3 + 1.5 * IQR

    dataframe[column] = np.where(
        dataframe[column] > upper_bound,
        upper_bound,
        dataframe[column]
    )

    dataframe[column] = np.where(
        dataframe[column] < lower_bound,
        lower_bound,
        dataframe[column]
    )

    return dataframe
    df = cap_outliers_iqr(df, 'Price')

df = cap_outliers_iqr(df, 'Mileage')

print("Outliers capped successfully!")
plt.figure(figsize=(8, 5))

sns.boxplot(x=df['Price'])

plt.title("Price After Outlier Treatment")

plt.show()
```

```python
numerical_cols = [
    'Year',
    'Mileage',
    'Engine_Size'
]

categorical_cols = [
    'Make',
    'Fuel_Type',
    'Transmission',
    'Condition'
]

print("Numerical Columns:", numerical_cols)

print("Categorical Columns:", categorical_cols)
```

```python
df_encoded = pd.get_dummies(
    df,
    columns=categorical_cols,
    drop_first=True
)

print("Categorical Data Converted Successfully!")

display(df_encoded.head())
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

df_encoded[numerical_cols] = scaler.fit_transform(
    df_encoded[numerical_cols]
)

print("Feature Scaling Completed Successfully!")

display(df_encoded.head())
```

```python
print("FINAL DATASET INFORMATION")

print("\nShape of Final Dataset:")
print(df_encoded.shape)

print("\nMissing Values:")
print(df_encoded.isnull().sum())

print("\nFirst 5 Rows:")
display(df_encoded.head())
```

```python
df_encoded.to_csv(
    'used_cars_preprocessed.csv',
    index=False
)

print("Final Preprocessed Dataset Saved Successfully!")
```
